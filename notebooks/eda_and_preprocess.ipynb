# Pré-processamento dos dados

Nesta etapa são removidas colunas que não serão utilizadas pelo modelo.

Foram mantidos apenas:

- título;
- texto;
- categoria.

Embora a coluna "text" contenha algumas notícias sem conteúdo, esses registros não foram descartados. Durante o treinamento, a entrada do modelo foi construída a partir da concatenação dos campos "title" e "text". Dessa forma, mesmo quando a coluna "text" da notícia está vazia, o título fornece informação suficiente para compor a entrada do modelo, evitando a existência de documentos vazios durante o treinamento.

```python
# Removendo as colunas que não serão considerados durante o treinamento 
df_filtered = df.drop(columns=['date', 'subcategory', 'link'])
df_filtered.head()
```

```python
print("Quantidade de valores ausentes por coluna:")
df_filtered.isna().sum()
```

```python
# Remove categorias com poucas amostras para reduzir o impacto de classes extremamente raras durante o treinamento
min_samples = 8
category_counts = df_filtered['category'].value_counts()
categories_to_keep = category_counts[category_counts >= min_samples].index
df_filtered = df_filtered[df_filtered['category'].isin(categories_to_keep)]
```

```python
# Mostrando Total de categorias e Total de itens no dataset filtrado
print(f"Dimensões do dataset após filtragem: {df_filtered.shape}")
print(f"Total de categorias únicas após filtragem: {df_filtered['category'].nunique()} \n")

# Verificando a distribuição de dados após a filtragem
print(f"Distribuição das categorias depois da filtragem:")
df_filtered['category'].value_counts()
```

```python
# Salvando o dataset filtrado
df_filtered.to_csv('../data/filtered_articles.csv', index=False)
```
